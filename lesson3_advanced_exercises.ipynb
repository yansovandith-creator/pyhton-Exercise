## Exercise 1: Truth table mini-lab

- Create booleans `p = True`, `q = False`.
- Print results for: `p and q`, `p or q`, `not p`, `not q`.
- Add one compound expression with parentheses and print its result.

```python
p = True
q = False

print(p and q)  # False
print(p or q)   # True
print(not p)    # False
print(not q)    # True

# Compound: exactly one of p, q is True (XOR)
print((p or q) and not (p and q))  # True
```

## Exercise 2: Comparison chain

- Given `a = 12`, `b = 12`, `c = 7`, print at least 6 comparison results.
- Include equality, inequality, greater/less, and a chained comparison (e.g., `c < a <= b`).

```python
a = 12
b = 7

print(a > b)    # True
print(a < b)    # False
print(a == b)   # False
print(a != b)   # True
print(a >= 12)  # True
```

## Exercise 3: Operator precedence check

- Create a complex boolean expression using `and`, `or`, and `not`.
- Evaluate it once without extra parentheses and once with explicit parentheses.
- Print both and comment whether they match.

```python
a, b, c = True, False, True

without_parens = a or b and not c
with_parens = a or (b and (not c))

print(without_parens)  # True
print(with_parens)     # True

# Match: Python evaluates not first, then and, then or,
# so the explicit parentheses show the same grouping.

# A different grouping gives a different result:
print((a or b) and (not c))  # False
```

## Exercise 4: Identity vs equality

- Create two equal lists and compare with `==` and `is`.
- Assign one list to another variable and compare again.
- Write a short comment explaining differences in output.

```python
list1 = [1, 2, 3]
list2 = [1, 2, 3]

print(list1 == list2)  # True  (same values)
print(list1 is list2)  # False (two separate objects in memory)

list3 = list1
print(list1 == list3)  # True
print(list1 is list3)  # True  (both names point to the same object)

# == compares contents; "is" compares object identity.
# list3 = list1 copies the reference, not the list.
```

## Exercise 5: Membership filter

- Given `items = ["python", "java", "go", "rust"]`, test membership for 4 queries.
- Use both `in` and `not in`.
- Print human-readable results like `python: found`.

```python
items = ["python", "java", "go", "rust"]

for query in ["python", "c++", "go", "ruby"]:
    if query in items:
        print(f"{query}: found")
    if query not in items:
        print(f"{query}: not found")
```
