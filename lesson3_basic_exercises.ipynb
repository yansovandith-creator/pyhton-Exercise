## Exercise 1: Boolean basics

## Exercise 2: Comparison operators

Given `a = 12` and `b = 7`, print the result of:

- `a > b`
- `a < b`
- `a == b`
- `a != b`
- `a >= 12`

```python
a = 12
b = 12
c = 7

print(a == b)       # True
print(a != c)       # True
print(a > c)        # True
print(c < b)        # True
print(a >= b)       # True
print(a <= c)       # False
print(c < a <= b)   # True (chained: c < a and a <= b)
```

## Exercise 3: Logical operators

1. Create `age = 20` and `has_id = True`.
2. Use `and` to check if someone can enter: age >= 18 and has_id is True.
3. Use `or` to check if `age < 18` or `has_id` is False.
4. Use `not` with `has_id` and print the result.

```python
a, b, c = True, False, True

without_parens = a or b and not c
with_parens = a or (b and (not c))

print(without_parens)  # True
print(with_parens)     # True

# Match: Python evaluates not first, then and, then or,
# so the explicit parentheses show the same grouping.

# A different grouping gives a different result:
print((a or b) and (not c))  # False
```

## Exercise 4: Identity and membership

1. Create two lists:
   - `list1 = [1, 2, 3]`
   - `list2 = list1`
2. Print `list1 is list2` and `list1 is not list2`.
3. Check membership in `list1`:
   - `2 in list1`
   - `5 not in list1`

```python
list1 = [1, 2, 3]
list2 = [1, 2, 3]

print(list1 == list2)  # True  (same values)
print(list1 is list2)  # False (two separate objects in memory)

list3 = list1
print(list1 == list3)  # True
print(list1 is list3)  # True  (both names point to the same object)

# == compares contents; "is" compares object identity.
# list3 = list1 copies the reference, not the list.
```

## Exercise 5: Arithmetic and assignment
